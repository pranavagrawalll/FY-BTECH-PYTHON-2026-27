```python
# Problem Statement: 
# To check whether or not the triangle is a right angle triangle using function.
```

```python
# Aim: 
# Write a Python Program that accepts the length of three sides of a triangle as inputs. The program should indicate whether or not the triangle is a right angle triangle using function.
```

```python
# Objective:
# To learn an implement functions
```

```python
# Case 1
def is_right_angled (a, b, c):
    sides=[a, b, c]
    sides.sort()
    a = sides[0]
    b = sides[1]
    c = sides[2]
    if a**2 + b**2 == c**2:
        return True
    else:
        return False

side1 = float(input("Enter first side:"))
side2 = float(input("Enter second side:"))
side3 = float(input("Enter third side:"))

if is_right_angled (side1, side2, side3):
    print("The Triangle is a right angled triangle.")
else:
    print("The Triangle is not a right angled triangle.")
```
