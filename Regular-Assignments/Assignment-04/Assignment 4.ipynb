```python
# Input:

# Program 1:

A = [
    [1,2],
    [3,4]
]
B= [
    [5,6],
    [7,8]
]
Result = [
    [0,0],
    [0,0]
]

for i in range (len(A)):
    for j in range (len(A[0])):
        Result[i][j] = A[i][j] + B[i][j]

print ("Matrix A:")
for row in A:
    print (row)

print ("Matrix B:")
for row in B:
    print (row)

print ("Addition of Matrices:")
for row in Result:
    print (row)
```

```python
# Program 2: (numpy)

import numpy as np

A=np.array ([
    [1,2],
    [3,4]
])

B=np.array ([
    [5,6],
    [7,8]
])

Result = A+B

print ("Matrix A:")
print (A)
print ("Matrix B:")
print (B)
print ("Addition of Matrices:")
print (Result)
```
