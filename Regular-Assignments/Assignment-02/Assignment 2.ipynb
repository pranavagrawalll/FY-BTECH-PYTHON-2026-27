```python
# Input:

#Ans. 

a = int(input("Enter 1st no.:"))
b = int(input("Enter 2nd no.:"))
c = int(input("Enter 3rd no.:"))

if a>=b and a>=c:
    print(a, "is the largest no.")
elif b>=a and b>=c:
    print (b, "is the largest no.")
else:
    print (c, "is the largest no.")
```
