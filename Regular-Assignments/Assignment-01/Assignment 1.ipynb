```python
# Input:

# Program to perform different operations on Dictionary, Tuple and a List data structure
# Step 1 : Start
# Step 2 : Create a Dictionary, Tuple and a List

student={
    "name" : "Pranav",
    "roll_no" : 373,
    "branch" : "CSE"
}
a_tuple = (10,20,30,40,20,50)
b_list = [10,20,30,40,50]
```

```python
# dict. operations
print ("Dictionary Operations")

# add a key value pair
student["marks"]=90
print ("After adding marks:", student)

# update a value
student["marks"]=95
print ("After updating marks:", student)

# Access a value
print ("Student's name:", student["name"])

# Delete a key value pair
del student["roll_no"]
print ("After deleting roll_no:", student)

# Display Keys and Values
print ("Keys:", student.keys())
print ("Values:", student.values())
```

```python
# Tuple Operations
print("Tuple Operations")

# Access element using indexing
print ("First element:", a_tuple[0])
print ("Third element:", a_tuple[2])

# Perform slicing
print ("Tuple Sclicing:", a_tuple[1:4])

# Find length
print ("Length of Tuple:", len(a_tuple))

# Find Max and Min
print ("Maximum Value:", max(a_tuple))
print ("Minimum Value:", min(a_tuple))

# Count an element
print ("Count of 20:", a_tuple.count(20))

# Find index of an element
print ("Index of 30:", a_tuple.index(30))
```

```python
# List Operations
print ("List Operations")

# add an element using append()
b_list.append(60)
print ("After append:", b_list)

# insert element using insert()
b_list.insert(1,15)
print ("Afetr inserting:", b_list)

# remove element using remove()
b_list.remove(15)
print ("After removing:", b_list)

# delete element using pop()
removed_elements = b_list.pop()
print ("Popped element:", removed_elements)
print ("After popping:", b_list)

# sort the list
b_list.sort()
print ("After sorting:", b_list)

# reverse the list
b_list.reverse()
print ("After reversing:", b_list)

# access element using indexing
print ("First Element:", b_list[0])

# access element using slicing
print ("List Slicing:", b_list[1:4])

# step 7: Stop
```
